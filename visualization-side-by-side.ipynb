```python
import os
import pathlib

def setup_credentials():
    credential_file = pathlib.Path.home() / ".aws/credentials"
    with credential_file.open() as f:
        lines = list(f)[1:]
        mapping = dict(line.split("=") for line in lines)

    os.environ["AWS_ACCESS_KEY_ID"] = mapping["aws_access_key_id"].strip()
    os.environ["AWS_SECRET_ACCESS_KEY"] = mapping["aws_secret_access_key"].strip()

setup_credentials()

os.environ.pop("PROJ_PATH", None)
```
